```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 10, 100)

growth = np.exp(0.3 * t)

plt.plot(t, growth)
plt.xlabel("TimeoutError")
plt.ylabel("Growth")
plt.title("Exponential Growth")
plt.show()
```
